```python
import json
weather_data = {
    "status": "success",
    "city": "Edinburgh",
    "country": "UK",
    "current": {
        "temperature": 18,
        "humidity": 65,
        "condition": "Cloudy",
        "wind": {
            "speed": 12,
            "direction": "West"
        }
    },
    "forecast": [
        {
            "day": "Monday",
            "temperature": 17,
            "condition": "Rain"
        },
        {
            "day": "Tuesday",
            "temperature": 20,
            "condition": "Sunny"
        },
        {
            "day": "Wednesday",
            "temperature": 16,
            "condition": "Cloudy"
        }
    ]
}
with open("weather_data.json", "w") as file:
    json.dump(weather_data, file)

def read_weather(filename):
    with open(filename, "r") as file:
        return json.load(file)
def current_weather(filename):
    weather_data = read_weather(filename)
    city= weather_data["city"]
    temperature=  weather_data["current"]["temperature"]
    contition = weather_data["current"]["condition"]
    humidity= weather_data["current"]["humidity"]
    return{
        "city":city ,
        "temperature":temperature,
        "condition":contition,
        "humidity":humidity
    }
print(current_weather("weather_data.json"))
def safe_acess(filename):
    weather_data = read_weather(filename)
    city= weather_data.get("city", "unknown city")
    temperature=  weather_data.get("current", {}).get("temperature", "unknown temp value")
    contition = weather_data.get("current", {}).get("condition", "unknown condition")
    humidity= weather_data.get("current", {}).get("humidity", " humidity value unkown")
    return{
        "city":city ,
        "temperature":temperature,
        "condition":contition,
        "humidity":humidity
    }
print(safe_acess("weather_data.json"))

def wind_report(filename):
    weather_data = read_weather(filename)
    wind = weather_data.get("current", {}).get("wind","nothing found in wind")
    for key,value in wind.items():
        print(f"{key}: {value}")
wind_report("weather_data.json")

def forecast_temp(filename):
    weather_data = read_weather(filename)
    temperature = []
    for temp in weather_data["forecast"]:
        temperature.append(temp["temperature"])
    return temperature
print(forecast_temp("weather_data.json"))

def highest_temp(filename):
    weather_data = read_weather(filename)
    forecast = weather_data["forecast"]
    high = max(forecast, key= lambda temp:temp["temperature"])
    return {
        "day": high["day"], 
        "temperature": high["temperature"]}
print(highest_temp("weather_data.json"))
def add_forecast(filename, day_data):
    weather_data = read_weather(filename)
    weather_data["forecast"].append(day_data)
    print(weather_data)
    with open(filename, "w") as file:
        json.dump(weather_data, file)
add_forecast("weather_data.json", {
    "day":"Thursday",
    "temperature":22,
    "condition":"Sunny"
})
def update_current_weather(filename, updates):
    weather_data= read_weather(filename)
    weather_data["current"].update(updates)
    print(weather_data)
    with open(filename, "w") as file:
        json.dump(weather_data, file)
update_current_weather("weather_data.json", {
"temperature":25,
"condition":"Sunny"
})
def weather_report(filename):
    weather_data= read_weather(filename)
    city = weather_data["city"]
    temperatures= forecast_temp(filename)
    average_forecast_temperature= sum(temperatures)/len(temperatures)
    highest_temperature= highest_temp(filename)["temperature"]
    forecast_days= len(weather_data["forecast"])
    conditions = {"Rain": 0,
                  "Sunny": 0,
                  "Cloudy": 0}
    for data in weather_data["forecast"]:
        condition = data["condition"]
        conditions[condition] = conditions.get(condition,0)+1

    return{
    "city": city,
    "average_forecast_temperature":average_forecast_temperature,
    "highest_temperature":highest_temperature,
    "forecast_days":forecast_days,
    "conditions": conditions
    }
print(weather_report("weather_data.json"))

```

```python


library = [
    {
        "id": 1,
        "title": "Python Crash Course",
        "author": "Eric Matthes",
        "copies": 4,
        "details": {
            "category": "Programming",
            "year": 2023
        }
    },
    {
        "id": 2,
        "title": "Deep Learning",
        "author": "Ian Goodfellow",
        "copies": 2,
        "details": {
            "category": "AI"
        }
    },
    {
        "id": 3,
        "title": "Clean Code",
        "author": "Robert Martin",
        "copies": 5
    }
]
#creating library.json file
with open("library.json", "w") as file:
    json.dump(library, file)
#reading file helper function
def read_file(filename):
    with open(filename, "r") as file:
        return json.load(file)
#file writing helper function
def write_file(filename, data):
    with open(filename, "w") as file:
        json.dump(data, file)
#adding new file
def add_book(filename, book):
    library = read_file(filename)
    library.append(book)
    write_file(filename, library)
add_book("library.json", {"id": 4,"title": "Coding with Agents","author": "Rohi David","copies": 3})
#searching book by author
def search_author(filename, author):
    library= read_file(filename)
    author_books= [books for books in library if books["author"]== author]
    return author_books
print(search_author("library.json","Eric Matthes"))
#check for category
def category(filename, id):
    library = read_file(filename)
    for books in library:
        if books["id"] ==  id:
            book_category = books.get("details",{}).get("category","Unknown Category")
            return book_category
    return "Book Id not found"
print(category("library.json", 2))
#adding default details
def add_detail(filename):
    library= read_file(filename)
    for books in library:
        books.setdefault("details", {})
    write_file(filename, library)
add_detail("library.json")
#updating book information 
def update_book(filename, id, information):
    library = read_file(filename)
    for book in library:
        if book["id"]== id:
            book.update(information)
    print(library)
    write_file(filename, library)
update_book("library.json", 2 , {"copies":5,"publisher":"MIT Press"})
#removing details
def remove_detail(filename, id, detail):
    library = read_file(filename)
    for book in library:
        if book["id"] == id:
            removed = book["details"].pop(detail, None)
            if removed is None:
                return "Detail not found"
    write_file(filename, library)
    return "Detail removed"
print(remove_detail("library.json", 1, "year"))

def details(filename):
    library = read_file(filename)
    for books in library:
        settings = books.get("details",{})
        for key,value in settings.items():
            print(f"{key} : {value}")
details("library.json")

def backup_book(filename, id, copies):
    library = read_file(filename)
    for book in library:
        if book["id"] == id:
            new_copy = book.copy()
            new_copy["copies"] = copies
            return {
                "original": book["copies"],
                "backup": new_copy["copies"]
            }
    return "Book Not found"
print(backup_book("library.json", 4,998))

#getting all statistics
def library_statistics(filename):
    library = read_file(filename)
    total_books = len(library)
    most_copies = max(library, key=lambda book:book["copies"])
    least_copies = min(library, key=lambda book:book["copies"])
    categories = [book.get("details", {}).get("category", "unknown") for book in library]
    authors = [book["author"] for book in library]

    return{
    "total_books": total_books,
    "most_copies": most_copies,
    "least_copies": least_copies,
    "categories": categories,
    "authors": authors
}
print(library_statistics("library.json"))
```
