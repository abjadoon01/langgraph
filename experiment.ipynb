```python
import ast
import math
operator_dict={
    ast.Add: lambda x,y:x+y,
    ast.Sub: lambda x,y:x-y,
    ast.Mult: lambda x,y:x*y,
    ast.Div: lambda x,y:x/y,
}
function_dict={
    "sqrt": math.sqrt,
    "sin" : math.sin,
    "cos": math.cos,
    "tan": math.tan,
    "log": math.log
}
def calculator(expression):
    try:
        tree = ast.parse(expression, mode="eval")
        validation = validate_expression(tree)
        if not validation["valid"]:
            return validation
        result = evaluate_node(tree.body)
        return {
            "Success" :True,
            "Result": result,
            "Error": None
        }
    except Exception as e:
        return {
            "Success" :False,
            "Result": None,
            "Error": f"Error: {str(e)}"
            }
def validate_expression(tree):
    for node in ast.walk(tree):
        if isinstance(node, ast.Expression):
            pass
        elif isinstance(node, ast.Constant):
            if not isinstance(node.value, (int, float)):
                return{
                    "valid": False,
                    "Error": "Only number input are accepted"
                }
        elif isinstance(node, ast.BinOp):
            if type(node.op) not in operator_dict:
                return{
                    "valid": False,
                    "Error": "Not a valid operator"
                    }
            elif isinstance(node, ast.UnaryOp):
                if type(node.op) not in (ast.UAdd, ast.USub):
                    return {
                        "valid": False,
                        "Error": "Not a valid unary operator"
                        }
            elif isinstance(node, (ast.UAdd, ast.USub)):
                    pass
        elif isinstance(node, ast.Call):
            if not isinstance(node.func, ast.Name):
                return{
                    "valid": False,
                    "Error": "Not a valid function name"
                    }
            elif node.func.id not in function_dict:
                return{
                    "valid": False,
                    "Error": "Not a valid function"
                    }
            elif isinstance(node, ast.Name):
                pass
            elif isinstance(node, ast.Load):
                pass
        else: 
            return{
                "valid": False,
                "Error": "Not valid expression"
                    }
    return{
        "valid": True,
        "Error": None
        }
def evaluate_node(node):
    try:
        if isinstance(node, ast.Constant):
            return node.value
        elif isinstance(node, ast.BinOp):
            left = evaluate_node(node.left)
            right = evaluate_node(node.right)
            operation = operator_dict[type(node.op)]
            return operation(left,right)
        elif isinstance(node, ast.Call):
            func = function_dict[node.func.id]
            arg = [evaluate_node(arg) for arg in node.args]
            return func(*arg)
    except Exception as e:
        return "str(e)"
expression= "2*3"
cal_results = calculator(expression)
print(cal_results)
```
